```python
import pandas as pd
from glob import glob

trip_folder = "/Users/saraseid/Desktop/citibike-weather-project/2022-citibike-tripdata"

csv_files = sorted(glob(trip_folder + "/*.csv"))
len(csv_files), csv_files[:3]
```

```python
bike = pd.concat((pd.read_csv(f) for f in csv_files), ignore_index=True)
bike.shape
```

```python
needed = ["start_station_name","end_station_name","start_lat","start_lng","end_lat","end_lng"]
bike = bike[needed].dropna()
bike.head()
```

```python
# Create trip count column
bike["value"] = 1

# Step 1 — Group ONLY by station names
routes = (
    bike.groupby(["start_station_name", "end_station_name"])
        .size()
        .reset_index(name="trips")
)

# Step 2 — Keep ONE coordinate pair per route
coords = (
    bike[["start_station_name","end_station_name","start_lat","start_lng","end_lat","end_lng"]]
        .drop_duplicates(subset=["start_station_name","end_station_name"])
)

# Step 3 — Merge counts with coordinates
routes = routes.merge(coords, on=["start_station_name","end_station_name"], how="left")

# Step 4 — Limit to top 500 busiest routes
routes_top500 = routes.sort_values("trips", ascending=False).head(500).reset_index(drop=True)

routes_top500.head()
```

I created separate point layers for start and end stations to show where trips begin and end. I then added an Arc layer connecting the stations to represent trip flows. I set the arc color based on the trips field so that darker arcs represent routes with higher usage. This makes it easy to visually identify the most frequently traveled routes between bike stations.

After applying a filter for the most frequent trips, the map shows a dense cluster of bike activity in New York City. The busiest routes are concentrated in a small urban area, indicating short trips between nearby stations. This suggests that bike-sharing is mainly used for commuting and quick travel in densely populated, well-connected parts of the city
